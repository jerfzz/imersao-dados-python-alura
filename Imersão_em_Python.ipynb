````python
import numpy as np

#Criação de um dataframe de teste para usar de exemplo
df_salarios = pd.DataFrame({
    "nome":["Ana", "Bruna", "Carlos", "Daniele", "Eduardo"],
    "salario":[4000, np.nan, 5000, np.nan, 100000]
})

#Calcula a media e substitui os nulos pela media
df_salarios["salario_media"] = df_salarios["salario"].fillna(df_salarios["salario"].mean().round(2)) #fillna: preencher os valores nulos

#Calcula a mediana e substitui os nulos pela mediana
df_salarios["salario_mediana"] = df_salarios["salario"].fillna(df_salarios["salario"].median())
````

````python
# 1. Seu filtro inicial
df_ds = df_limpo[df_limpo['cargo'] == 'Data Scientist']

# 2. Agrupamento por residência (país)
df_medio = df_ds.groupby('residencia')['usd'].mean().reset_index()
df_medio.head(15)

# 3. CRIAR AS FAIXAS: Definimos os cortes de 100k em 100k
bins = [0, 100000, 200000, 300000, 400000, 500000]
labels = ['Até 100k', '100k - 200k', '200k - 300k', '300k - 400k', '400k+']

# Criamos a coluna de faixa baseada na média de cada país
df_medio['faixa_salarial'] = pd.cut(df_medio['usd'], bins=bins, labels=labels, right=False)

# 4. Ordenar para o funil ficar correto
df_medio = df_medio.sort_values('usd', ascending=False)


# 5. Gerar o gráfico
# DICA: Usando 'faixa_salarial' no color, o Plotly agrupa visualmente os países do mesmo nível
fig3 = px.funnel(df_medio,
                 x='usd',
                 y='residencia', # Mantém os países no eixo Y
                 color='faixa_salarial', # Pinta os países conforme a faixa de 100k
                 title='Salário Médio por País e Faixa - Data Scientist',
                 labels={'usd': 'Média Salarial (USD)', 'residencia': 'País', 'faixa_salarial': 'Nível'},
                 color_discrete_sequence=px.colors.sequential.Viridis)

fig3.show()
````

````python
import pandas as pd
import plotly.express as px

# 1. Filtra o cargo
df_ds = df_limpo[df_limpo['cargo'] == 'Data Scientist']

# 2. Calcula a média por país
df_medio = df_ds.groupby('residencia')['usd'].mean().reset_index()

# 3. Cria as faixas de 100k (FAÇA ISSO ANTES DE CORTAR O DF)
bins = [0, 100000, 200000, 300000, 400000, 1000000] # 1M para garantir o teto
labels = ['Até 100k', '100k - 200k', '200k - 300k', '300k - 400k', '400k+']
df_medio['faixa_salarial'] = pd.cut(df_medio['usd'], bins=bins, labels=labels, right=False)

# 4. Ordena pelos maiores salários
df_medio = df_medio.sort_values('usd', ascending=False)

# 5. PEGA OS 15 MELHORES (Agora sim!)
df_top15 = df_medio.head(15)

# 6. Gera o gráfico usando o df_top15
fig3 = px.funnel(df_top15,
                 x='usd',
                 y='residencia',
                 title='Top 15 Países - Salário Médio Data Scientist',
                 labels={'usd': 'Média Salarial (USD)', 'residencia': 'País', 'faixa_salarial': 'Nível'},
                 color='faixa_salarial',
                 color_discrete_sequence=px.colors.sequential.Viridis)

fig3.show()
````

````python
# 1. Criar o agrupamento com média e contagem (tamanho da amostra)
df_bubble = df_ds.groupby('residencia').agg(
    media_usd=('usd', 'mean'),
    quantidade=('usd', 'count')
).reset_index()

# 2. Criar o gráfico
fig_bubble = px.scatter(df_bubble,
                        x='media_usd',
                        y='quantidade',
                        size='quantidade',
                        color='media_usd',
                        hover_name='residencia',
                        text='residencia', # Nome do país em cima da bolha
                        title='Salário Médio vs. Número de Profissionais',
                        labels={'media_usd': 'Salário Médio (USD)', 'quantidade': 'Total de Registros'},
                        color_continuous_scale='Viridis',
                        size_max=60)

# Ajustar posição do texto para não ficar em cima da bolha
fig_bubble.update_traces(textposition='top center')

fig_bubble.show()
````

````python
# Agrupando para ter volume e média
df_tree = df_ds.groupby('residencia').agg(
    media_usd=('usd', 'mean'),
    contagem=('usd', 'count')
).reset_index()

fig = px.treemap(df_tree,
                 path=['residencia'],
                 values='contagem',
                 color='media_usd',
                 title='Volume de Vagas (Tamanho) vs. Salário Médio (Cor)',
                 color_continuous_scale='RdBu_r')
fig.show()
````

### 1. Treemap (Hierarquia e Proporção)

O Treemap é excelente para substituir o gráfico de Pizza quando você tem muitos países. Ele organiza os países em blocos: o **tamanho** do bloco pode ser a quantidade de vagas e a **cor** o salário médio.

```python
# Agrupando para ter volume e média
df_tree = df_ds.groupby('residencia').agg(
    media_usd=('usd', 'mean'),
    contagem=('usd', 'count')
).reset_index()

fig = px.treemap(df_tree,
                 path=['residencia'],
                 values='contagem',
                 color='media_usd',
                 title='Volume de Vagas (Tamanho) vs. Salário Médio (Cor)',
                 color_continuous_scale='RdBu_r')
fig.show()

```

---

### 2. Sunburst (Exploração Multinível)

Se você quiser adicionar uma segunda dimensão (ex: **Nível de Experiência**), o Sunburst é imbatível. Ele mostra o país no círculo interno e abre as fatias de experiência no círculo externo.

```python
# Supondo que você tenha uma coluna 'experiencia'
fig = px.sunburst(df_ds,
                  path=['residencia', 'experiencia'],
                  values='usd',
                  title='Hierarquia de Salários: País e Experiência',
                  color='usd',
                  color_continuous_scale='Viridis')
fig.show()

```

---

### 3. Gráfico de Violino (A "Evolução" do Boxplot)

Como você já fez o Boxplot, o Violino é o próximo passo. Ele mostra a **curva de densidade**, ou seja, onde a maioria dos salários está "inchada". É visualmente muito elegante e mostra se a distribuição é unimodal (um pico) ou bimodal (dois picos de salário no mesmo país).

```python
fig = px.violin(df_ds,
                y="usd",
                x="residencia",
                color="residencia",
                box=True,
                points="all",
                title="Distribuição Detalhada (Densidade) por País")
fig.show()

```

---

### Qual escolher?

* **Treemap:** Se quiser mostrar quais países "dominam" o mercado em quantidade.
* **Sunburst:** Se quiser cruzar país com outra variável (Seniority, Tipo de Contrato).
* **Violino:** Se quiser um visual moderno que mostre a "massa" de onde as pessoas ganham mais.

````python
import pycountry

# Função para converter ISO-2 para ISO-3
def iso2_to_iso3(code):
    try:
        return pycountry.countries.get(alpha_2=code).alpha_3
    except:
        return None

# Criar nova coluna com código ISO-3
df_limpo["residencia_iso3"] = df_limpo["residencia"].apply(iso2_to_iso3)

# Calcular média salarial por país (ISO-3)
df_ds = df_limpo[df_limpo["cargo"] == "Data Scientist"]
media_ds_pais = df_ds.groupby("residencia_iso3")["usd"].mean().reset_index()

# Gerar o mapa
fig = px.choropleth(media_ds_pais,
                    locations="residencia_iso3",
                    color="usd",
                    color_continuous_scale="rdylgn",
                    title="Salário Médio de Cientista de Dados por País",
                    labels={"usd": "Salário Médio (USD)", "residencia_iso3": "País" })

fig.show()
````
